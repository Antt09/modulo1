```
import cv2

# Inicia a captura da webcam (o número 0 indica a primeira câmera detectada)
webcam = cv2.VideoCapture(0)

# Verificação de segurança: a câmera abriu?
if not webcam.isOpened():
    print("Erro: Não foi possível acessar a webcam")
    exit()

print("Webcam conectada com sucesso! Pressione 'q' na janela do vídeo para sair.")

while True:
    # Captura frame por frame da tela
    sucesso, frame = webcam.read()

    if not sucesso:
        print("Erro ao receber o frame da câmera.")
        break

    # Exibe o frame em uma janela chamada 'Teste webcam'
    cv2.imshow('Teste webcam', frame)

    # Espera 1 milissegundo. Se a tecla 'q' for pressionada, para o loop
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Libera a câmera para o sistema operacional e fecha as janelas
webcam.release()
cv2.destroyAllWindows()

```
